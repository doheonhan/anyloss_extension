```python
import bayesiantests as bt
import matplotlib.pyplot as plt

#https://matplotlib.org/stable/gallery/lines_bars_and_markers/horizontal_barchart_distribution.html
def stacked_bar(results, category_names):
    """
    Parameters
    ----------
    results : dict
        A mapping from question labels to a list of answers per category.
        It is assumed all lists contain the same number of entries and that
        it matches the length of *category_names*.
    category_names : list of str
        The category labels.
    """
    labels = list(results.keys())
    data = np.array(list(results.values()))
    data_cum = data.cumsum(axis=1)
    category_colors = plt.colormaps['RdYlGn'](
        np.linspace(0.15, 0.85, data.shape[1]))

    fig, ax = plt.subplots(figsize=(9.2, 5))
    ax.invert_yaxis()
    ax.xaxis.set_visible(False)
    ax.set_xlim(0, np.sum(data, axis=1).max())

    for i, (colname, color) in enumerate(zip(category_names, category_colors)):
        widths = data[:, i]
        starts = data_cum[:, i] - widths
        rects = ax.barh(labels, widths, left=starts, height=0.5,
                        label=colname, color=color)

        r, g, b, _ = color
        text_color = 'white' if r * g * b < 0.5 else 'darkgrey'
        ax.bar_label(rects, label_type='center', color=text_color)
    ax.legend(ncols=len(category_names), bbox_to_anchor=(0, 1),
              loc='upper left', fontsize='small')

    return fig, ax
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('5CV_MLP(WBCE+_SPL+)_102datasets.csv')
df
```

```python
df.rename(columns={'Unnamed: 0':'Metrics'}, inplace=True)
df
```

```python
# WBCEFL
df_f1 = df[df['Metrics'] == 'F1']
df_f1 = df_f1.drop('Dataset', axis=1)
df_f1 = df_f1.reset_index(drop=True)
df_wbcefl = df_f1.iloc[:,6:11].astype(float)
df_wbcefl
```

```python
# --- Row-wise normalization (Z-score) ---
# df_norm = df_wbcefl.sub(df_wbcefl.min(axis=1), axis=0).div(df_wbcefl.max(axis=1) - df_wbcefl.min(axis=1), axis=0)
df_norm = df_wbcefl.sub(df_wbcefl.mean(axis=1), axis=0).div(df_wbcefl.std(axis=1), axis=0)
df_norm
```

```python
# plt.rcParams["figure.figsize"] = [5, 3]
plt.rcParams["figure.dpi"] = 200
plt.rcParams['axes.titlesize'] = 20  
plt.rcParams['axes.linewidth'] = 2
plt.rcParams['axes.labelsize'] = 20  
plt.rcParams['font.size'] = 15

data = np.array(df_norm)  # your normalized array

# If df_norm might contain NaN/inf, sanitize it:
data = np.array(df_norm, dtype=float)
data = np.where(np.isfinite(data), data, np.nan)

# Use nanmean to ignore any NaNs produced by Z-score rows with std=0
mean_values = np.nanmean(data, axis=0)

x_labels = ["0.1", "0.3", "0.5", "0.7", "0.9"]
x = np.arange(len(x_labels))

# plt.figure(figsize=(8, 6))
cmap = plt.colormaps.get_cmap('tab20')

for i in range(data.shape[0]):
    color = cmap(i % cmap.N)
    plt.plot(x, data[i, :], color=color, alpha=0.4, linewidth=1)

# Draw the mean line on top
plt.plot(x, mean_values, color='black', linewidth=3, label='Average of 102', zorder=10)

plt.xticks(x, x_labels)
plt.xlabel("Hyperparam 'r'")
plt.ylabel('Z-Score')
plt.axhline(0, 0.01, 0.99, color='dimgrey', linestyle='--', linewidth=2.0)
plt.title("Normalized F1 with different 'r'")
plt.legend(loc='lower center')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
